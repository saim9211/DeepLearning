#### for random tensor creation bcz most of the machine learning used the random tensor  first creat the random tensor and then according to data update it optimize them
start with the random number -> look at the data-> update the random tensor -> look at the data -> update random tensor


```python
ran_Tensor=torch.rand(3,4)
print(ran_Tensor)
print(ran_Tensor.ndim)
print(ran_Tensor.shape)
# to get the type of the tensor
print(ran_Tensor.dtype)
```

```python
# it can adjust the size to be whatever we want
# it means 4 rows and 4 column 2 dimesional one matrix
torch.random=torch.rand(size=(4,4))
# it means 4 sample of 4 rows and 2 column
torch.random=torch.rand(size=(4,4,2))
# it means 2  sample(  (4sample(4,2))
# (2 batch size  (((4 channel  (4 rows,2  column))))
random_tensor=torch.rand(size=(2,4,4,2))
print(random_tensor.shape)
print("dimension:",random_tensor.ndim)
print(random_tensor)
```
