### reshaping ,stacking ,squeezing ,unsqueezing  tensors
1. torch.reshape(input, shape)	Reshapes input to shape (if compatible), can also use torch.Tensor.reshape().
2. Tensor.view(shape)	Returns a view of the original tensor in a different shape but shares the same data as the original tensor.
3. torch.stack(tensors, dim=0)	Concatenates a sequence of tensors along a new dimension (dim), all tensors must be same size.
4. torch.squeeze(input)	Squeezes input to remove all the dimenions with value 1.
5. torch.unsqueeze(input, dim)	Returns input with a dimension value of 1 added at dim.
6. torch.permute(input, dims)	Returns a view of the original input with its dimensions permuted (rearranged) to dims.

| Operation | What it does                         | Common use                                  |
| --------- | ------------------------------------ | ------------------------------------------- |
| reshape   | Changes shape safely                 | Flattening, refactoring tensor shape        |
| view      | Changes shape fast, needs contiguity | Simple reshaping when memory layout is fine |
| squeeze   | Removes size-1 dimensions            | Remove extra batch/channel dim              |
| unsqueeze | Adds size-1 dimensions               | Add batch dimension                         |
| permute   | Reorders axes                        | Image/video tensor format changes           |

```python
# stacking tensor on top of each other
# but the thing that remain in mind is that dim remain same never get change remain same as compared to original one
x_stacked=torch.stack([x,x,x,x],dim=1)
print(x_stacked)

print(x_stacked.shape)
```

```python
# squeezing means removing the dimension from tensor
x_org=torch.arange(1,9)
print(x_org)
print(x_org.shape)

x_reshaped=x_org.reshape(1,8)
print(x_reshaped)
print(x_reshaped.shape)

# now squeez that tensor
x_squeezed=x_reshaped.squeeze()
print(x_squeezed)
print("shape aftert the squeez",x_squeezed.shape)


#unsqeez means adding the dimension in to tensor
x_unsqueezed=x_squeezed.unsqueeze(dim=1)
print(x_unsqueezed)
print("shape before the squeez",x_squeezed.shape,"shape after the squeez",x_unsqueezed.shape)

```

```python
# permute it does not chnage the shape number wise ,it changes the axis wise
pic_tensor=torch.rand(size=(224,224,3)) # like height,width,colour-side
per_pic=pic_tensor.permute(2,0,1)# now colour-side,height,width
print("shape before the permute",pic_tensor.shape)
print("shape after the permute",per_pic.shape)
```

```python
print("values of original",pic_tensor[0,0,0])
print("values of permute",per_pic[0,0,0])
print("before change",pic_tensor[0,0,0])
pic_tensor[0,0,0]=1.222
print("after changing the values in the  original",pic_tensor[0,0,0]," also cause change in the permute",per_pic[0,0,0])
```

pytorch tensors and numpy
numply is popular scientigic python numerical computing library and bcz of pytorch it has functionality to interact with it.

1. data in numpy want in pytorch tensor -> torch.from_numpy()
2. pytorch tensor -> numpy -> torch.tensor.numpy()

```python
# numpy array to tensor
import torch
import numpy as np
array=np.arange(1.0,8.0)
print(array,"also the type",array.dtype)
# converting them in to tensor
tensor=torch.from_numpy(array)# when converting numpy to tensor  -> pytorch reflect numpy by default data type of float 64 unless specified otherwise
print(tensor,"also the type",tensor.dtype)
```

```python
#change the vlaue of the array and what effect on the tensor ?
array=array+1 # so keep in mind keep change in the array never causes the change in the tensor
array,tensor
```

```python
# tensor to numpy array
tensor=torch.ones(7)
numpy=tensor.numpy()# same numpy taking the by deafualt data type of the tensor
tensor,numpy
```

```python
# same changing in the value of the tensor never cause the change in the numpy
tensor=tensor+1
tensor,numpy
```

Reproducibility means you intentionally control randomness so the same code can produce the same random tensors, the same initialization, and the same training behavior again

```python
tensor_A=torch.rand(3,4)
tensor_B=torch.rand(3,4)
print(tensor_A)
print(tensor_B)
# generates the diff random number
print(tensor_A==tensor_B)
```

```python
# let make the some reproducible tensor
import torch

# set the random value
random_seed=42
torch.manual_seed(random_seed)
tensor_C=torch.rand(3,4)
# if yu creating the random seed one after one other so every time we had to caal teh torch.manual_seed()
torch.manual_seed(random_seed)
tensor_D=torch.rand(3,4)
print(tensor_C)
print(tensor_D)
print(tensor_C==tensor_D)
```
