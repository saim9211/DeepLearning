### tensor data types issue
tensor datatype  is one of the 3 big errors you will run in to with pytorch and deeplearning:
1. tensors not right data type
2. tensor not right shape
3. tensor not on right device

```python
float_32_tensor=torch.tensor([2.0,3.0,4.0],
                             dtype=None,
                             device=None, # on which device is your tensor on
                             requires_grad=False # whether or not to track gradients with this tensors operations
                             )
```

```python
# changing the data type
float_16_tensor=float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_32_tensor*float_32_tensor
```

```python
int_32_tensor=torch.tensor([3,6,9],dtype=torch.int64)
int_32_tensor
```

```python
float_32_tensor*int_32_tensor
```

####manipulating the tensor (tensor operation)

```python
# create
import torch
import numpy as np
import pandas as pd

tensor=torch.tensor([1,2,3])
# the main thing it only make changes one time and not make change in the original one
print(tensor+10)
print(tensor+100)
print(tensor+10)
```

```python
# sub
tensor=torch.tensor([1,2,3])
print(tensor-10)
```

```python
# builtin functions
tensor=torch.tensor([1,2,3])
print(torch.mul(tensor,10))
print(torch.add(tensor,10))
print(torch.sub(tensor,10))
print(torch.div(tensor,2))
```

### matrix multiplication in the pytorch

the number of column of 1 matrix will be equal to the number of rows of the 2 matrix
and the number of rows of 1 matrix will be equall to the number of the columns of the 2 matrix

1. the inner dimension must match for multiplication

(2,3)@(3,2) -> (2,2)


(3,2)@(2,3)-> (3,3)
